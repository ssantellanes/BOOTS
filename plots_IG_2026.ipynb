```python
#data.iloc[0,8][11:] '2016/12/31'
#data.iloc[0,8][:10] #'2010/01/01'#
import altair as alt
start=[]
end=[]
for i in range(len(data.iloc[:,8])):
    start.append(data.iloc[i,8][:10])
    end.append(data.iloc[i,8][11:])
data["Start"]=pd.Series(start)
data["End"]=pd.Series(end)
data["Start"]=pd.to_datetime(data["Start"])
data["End"]=pd.to_datetime(data["End"])
data["Station ID"]=data["Station ID"].astype(str)
bars = alt.Chart(data).mark_bar(color='#CC5500').encode(
    x="Start",
    x2="End",
    y=alt.Y("Station ID")
)

text = bars.mark_text(
    align="left",
    baseline="middle",
    color="white",
    fontSize=14,
    dx=5  # Nudges text to right so it doesn't appear on top of the bar
).encode(
    text="Station ID"
)
(bars+text).properties(height=400, width=600).configure_axis(
    labelFontSize=14,
    titleFontSize=20
)
```

```python
import seaborn as sns
grid=pygmt.datasets.load_earth_relief(resolution="05m",region=[100,300,-60,70])

fig=pygmt.Figure()
with pygmt.clib.Session() as session:
    session.call_module('gmtset', 'FONT 18p')
    session.call_module('gmtset', 'MAP_FRAME_TYPE fancy')
#fig.coast(projection="N12c",region=[120, 300, -40, 60],land="wheat4", water="lightcyan2",frame="afg")
seconds=120
season="Winter"
fig.coast(
    region=[100,300,-60,70],
    projection="Cyl_stere/12c",
    land="darkgray",
    water="white",
    borders="1/0.5p",
    shorelines="1/0.5p",
    frame="ag",
)

fig.grdimage(grid=grid,region=[100,300,-60,70],cmap="geo")
if seconds==120:   
    pygmt.makecpt(cmap="plasma",series=[-55,-37])
elif seconds==250:   
    pygmt.makecpt(cmap="plasma",series=[-52,-36])
elif seconds==800:   
    pygmt.makecpt(cmap="plasma",series=[-43,-38])
fig.plot(x=data.Lon,y=data.Lat,fill=data[f"{season} Power {seconds} s"],cmap=True,style="c0.45c",pen="black")
fig.colorbar(frame="af+lMedian Power dB")
#with fig.inset(position="jBL+w2.9c/1.2c+o4.3c/0c", box="+p0.5+gwhite"):
#    pass

#fig.plot(x=legend.lon, y=legend.lat,fill=legend.mag, pen='black', cmap=True,style='c0.45c')
#fig.text(text=['-50dB'], x=180, y=-57, font='9p,black')
#fig.text(text=['-45dB'], x=195, y=-57, font='9p,black')
##fig.text(text=['-40dB'], x=210, y=-57, font='9p,black')
#fig.text(text=['Legend'], x=195, y=-46, font='10p,black')
#fig.show()

fig.savefig(f"/Users/seansantellanes/Documents/Background_Spectra/figures/{season}_median_power_{seconds}_s.pdf")
```
